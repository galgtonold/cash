# Cash File Dependency Tracking Demo

This notebook demonstrates the automatic file dependency tracking feature of `cash`. When you read a file, `cash` automatically tracks it as a dependency and invalidates the cache if the file changes.

```python
%cash_on
```

```python
%cash_debug on
```

## 1. Standard File I/O (`open`)

```python
# Create a test file
with open('test_data.txt', 'w') as f:
    f.write('Version 1')
```

```python
# Read the file. This should be cached.
with open('test_data.txt', 'r') as f:
    content = f.read()
print(f"Read content: {content}")
import time 
time.sleep(2)
```

```python
# Rerunning this cell should hit the cache (check logs if debug is on)
print(content)
```

```python
# Modify the file
import time
time.sleep(1.1) 
with open('test_data.txt', 'w') as f:
    f.write('Version 3')
```

```python
# Rerunning the read cell should now RE-EXECUTE automatically because the file changed.
with open('test_data.txt', 'r') as f:
    content = f.read()
print(f"Read content: {content}")
```

## 2. Pandas (`read_csv`)

```python
import pandas as pd
df = pd.DataFrame({'a': [1, 2], 'b': [3, 4]})
df.to_csv('test_data.csv', index=False)
```

```python
df_read = pd.read_csv('test_data.csv')
print(df_read)
```

```python
# Modify CSV
time.sleep(1.1)
df2 = pd.DataFrame({'a': [10, 20], 'b': [30, 40]})
df2.to_csv('test_data.csv', index=False)
```

```python
# Should re-execute and show new data
df_read = pd.read_csv('test_data.csv')
print(df_read)
```

## 3. Pathlib

```python
from pathlib import Path
Path('test_path.txt').write_text('Pathlib V1')
```

```python
text = Path('test_path.txt').read_text()
print(text)
```

```python
time.sleep(1.1)
Path('test_path.txt').write_text('Pathlib V2')
```

```python
# Should show Pathlib V2
text = Path('test_path.txt').read_text()
print(text)
```

## 4. Numpy

```python
import numpy as np
np.savetxt('test_numpy.txt', [1, 2, 3])
```

```python
arr = np.loadtxt('test_numpy.txt')
print(arr)
```
