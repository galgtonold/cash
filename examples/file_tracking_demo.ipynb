# Which file reads cash tracks

cash tracks the common ways to read a file: `open()`, `pathlib`, pandas and
numpy readers. Each section below follows the same steps:

1. write a small file;
2. read it, then read it again in the next cell: the second read says CACHED;
3. change the file and read it once more: that read says EXECUTED and shows
   the new content.

The files go to a folder under your system's temporary directory.

**Read the badge, not the clock.** cash prints a badge under every cell it runs.
Open it to see one row per statement: **CACHED** (green) means cash restored the
value instead of running the code, **EXECUTED** (ochre) means the code ran.
Printed output is no evidence either way: a restored statement replays what it
printed.

```python
import cash
%cash_on
```

```python
import tempfile
from pathlib import Path

import numpy as np
import pandas as pd

folder = Path(tempfile.gettempdir()) / "cash_file_tracking_demo"
folder.mkdir(exist_ok=True)
text_file = folder / "notes.txt"
csv_file = folder / "table.csv"
numbers_file = folder / "numbers.txt"
```

## 1. `open()`

```python
with open(text_file, "w", encoding="utf-8") as f:
    f.write("Version 1")
```

```python
with open(text_file, encoding="utf-8") as f:
    content = f.read()
print(content)
```

```python
# The same read again: CACHED, because the file is unchanged.
with open(text_file, encoding="utf-8") as f:
    content = f.read()
print(content)
```

```python
with open(text_file, "w", encoding="utf-8") as f:
    f.write("Version 2")
```

```python
# The file changed, so this read runs: EXECUTED, and it prints Version 2.
with open(text_file, encoding="utf-8") as f:
    content = f.read()
print(content)
```

## 2. pandas

```python
pd.DataFrame({"a": [1, 2], "b": [3, 4]}).to_csv(csv_file, index=False)
```

```python
table = pd.read_csv(csv_file)
table
```

```python
table = pd.read_csv(csv_file)  # unchanged: CACHED
table
```

```python
pd.DataFrame({"a": [10, 20], "b": [30, 40]}).to_csv(csv_file, index=False)
```

```python
table = pd.read_csv(csv_file)  # changed: EXECUTED
table
```

## 3. `pathlib`

```python
text_file.write_text("Pathlib 1", encoding="utf-8")
```

```python
text = text_file.read_text(encoding="utf-8")
print(text)
```

```python
text = text_file.read_text(encoding="utf-8")  # unchanged: CACHED
print(text)
```

```python
text_file.write_text("Pathlib 2", encoding="utf-8")
```

```python
text = text_file.read_text(encoding="utf-8")  # changed: EXECUTED
print(text)
```

## 4. numpy

```python
np.savetxt(numbers_file, [1, 2, 3])
```

```python
numbers = np.loadtxt(numbers_file)
print(numbers)
```

```python
numbers = np.loadtxt(numbers_file)  # unchanged: CACHED
print(numbers)
```

```python
np.savetxt(numbers_file, [10, 20, 30])
```

```python
numbers = np.loadtxt(numbers_file)  # changed: EXECUTED
print(numbers)
```

The full list of tracked readers, and how cash decides a file changed, is in
[What counts as a change](https://cash-lib.readthedocs.io/en/latest/how-it-works/invalidation/#what-counts-as-a-change).
