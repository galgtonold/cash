```python
import pandas as pd
import numpy as np
import os
import time
import urllib.request
import gzip
import shutil
```

```python
# ── Cell 3: Download IMDb datasets ──
_data_dir = os.path.join(os.getcwd(), 'examples', 'large_scale_projects', 'data', 'imdb')
os.makedirs(_data_dir, exist_ok=True)

_imdb_files = {
    'title.basics': 'https://datasets.imdbws.com/title.basics.tsv.gz',
    'title.ratings': 'https://datasets.imdbws.com/title.ratings.tsv.gz',
    'name.basics': 'https://datasets.imdbws.com/name.basics.tsv.gz',
    'title.principals': 'https://datasets.imdbws.com/title.principals.tsv.gz',
    'title.crew': 'https://datasets.imdbws.com/title.crew.tsv.gz',
}

_opener = urllib.request.build_opener()
_opener.addheaders = [('User-Agent', 'Mozilla/5.0 (Cash Stress Test)'), ('Accept-Encoding', 'identity')]
urllib.request.install_opener(_opener)

_t0_dl = time.time()
_download_results = []

# Download each file individually (avoiding for-loop + comprehension Issue 24)
_fname = 'title.basics'
_url = _imdb_files[_fname]
_gz_path = os.path.join(_data_dir, _fname + '.tsv.gz')
_tsv_path = os.path.join(_data_dir, _fname + '.tsv')
if not os.path.exists(_tsv_path):
    if not os.path.exists(_gz_path):
        print(f'Downloading {_fname}...')
        urllib.request.urlretrieve(_url, _gz_path)
    print(f'Extracting {_fname}...')
    with gzip.open(_gz_path, 'rb') as _fin:
        with open(_tsv_path, 'wb') as _fout:
            shutil.copyfileobj(_fin, _fout)
_sz = os.path.getsize(_tsv_path) / 1e6
_download_results.append((_fname, _sz))
print(f'  {_fname}: {_sz:.1f} MB')

_fname = 'title.ratings'
_url = _imdb_files[_fname]
_gz_path = os.path.join(_data_dir, _fname + '.tsv.gz')
_tsv_path = os.path.join(_data_dir, _fname + '.tsv')
if not os.path.exists(_tsv_path):
    if not os.path.exists(_gz_path):
        print(f'Downloading {_fname}...')
        urllib.request.urlretrieve(_url, _gz_path)
    print(f'Extracting {_fname}...')
    with gzip.open(_gz_path, 'rb') as _fin:
        with open(_tsv_path, 'wb') as _fout:
            shutil.copyfileobj(_fin, _fout)
_sz = os.path.getsize(_tsv_path) / 1e6
_download_results.append((_fname, _sz))
print(f'  {_fname}: {_sz:.1f} MB')

_fname = 'name.basics'
_url = _imdb_files[_fname]
_gz_path = os.path.join(_data_dir, _fname + '.tsv.gz')
_tsv_path = os.path.join(_data_dir, _fname + '.tsv')
if not os.path.exists(_tsv_path):
    if not os.path.exists(_gz_path):
        print(f'Downloading {_fname}...')
        urllib.request.urlretrieve(_url, _gz_path)
    print(f'Extracting {_fname}...')
    with gzip.open(_gz_path, 'rb') as _fin:
        with open(_tsv_path, 'wb') as _fout:
            shutil.copyfileobj(_fin, _fout)
_sz = os.path.getsize(_tsv_path) / 1e6
_download_results.append((_fname, _sz))
print(f'  {_fname}: {_sz:.1f} MB')

_fname = 'title.principals'
_url = _imdb_files[_fname]
_gz_path = os.path.join(_data_dir, _fname + '.tsv.gz')
_tsv_path = os.path.join(_data_dir, _fname + '.tsv')
if not os.path.exists(_tsv_path):
    if not os.path.exists(_gz_path):
        print(f'Downloading {_fname}...')
        urllib.request.urlretrieve(_url, _gz_path)
    print(f'Extracting {_fname}...')
    with gzip.open(_gz_path, 'rb') as _fin:
        with open(_tsv_path, 'wb') as _fout:
            shutil.copyfileobj(_fin, _fout)
_sz = os.path.getsize(_tsv_path) / 1e6
_download_results.append((_fname, _sz))
print(f'  {_fname}: {_sz:.1f} MB')

_fname = 'title.crew'
_url = _imdb_files[_fname]
_gz_path = os.path.join(_data_dir, _fname + '.tsv.gz')
_tsv_path = os.path.join(_data_dir, _fname + '.tsv')
if not os.path.exists(_tsv_path):
    if not os.path.exists(_gz_path):
        print(f'Downloading {_fname}...')
        urllib.request.urlretrieve(_url, _gz_path)
    print(f'Extracting {_fname}...')
    with gzip.open(_gz_path, 'rb') as _fin:
        with open(_tsv_path, 'wb') as _fout:
            shutil.copyfileobj(_fin, _fout)
_sz = os.path.getsize(_tsv_path) / 1e6
_download_results.append((_fname, _sz))
print(f'  {_fname}: {_sz:.1f} MB')

_dl_elapsed = time.time() - _t0_dl
_total_sz = sum(_r[1] for _r in _download_results)
print(f'\nTotal: {_total_sz:.0f} MB in {_dl_elapsed:.1f}s')
```

```python
# ── Cell 4: Load datasets ──
_t0_load = time.time()

_titles = pd.read_csv(
    os.path.join(_data_dir, 'title.basics.tsv'),
    sep='\t', na_values='\\N', low_memory=False,
    dtype={'tconst': str, 'titleType': str, 'primaryTitle': str, 'startYear': str}
)
print(f'title.basics: {len(_titles):,} rows, {_titles.memory_usage(deep=True).sum()/1e6:.0f} MB')

_ratings = pd.read_csv(
    os.path.join(_data_dir, 'title.ratings.tsv'),
    sep='\t', na_values='\\N'
)
print(f'title.ratings: {len(_ratings):,} rows')

_names = pd.read_csv(
    os.path.join(_data_dir, 'title.basics.tsv').replace('title.basics', 'name.basics'),
    sep='\t', na_values='\\N', low_memory=False,
    dtype={'nconst': str, 'primaryName': str}
)
print(f'name.basics: {len(_names):,} rows, {_names.memory_usage(deep=True).sum()/1e6:.0f} MB')

_principals = pd.read_csv(
    os.path.join(_data_dir, 'title.principals.tsv'),
    sep='\t', na_values='\\N', low_memory=False,
    dtype={'tconst': str, 'nconst': str, 'category': str}
)
print(f'title.principals: {len(_principals):,} rows, {_principals.memory_usage(deep=True).sum()/1e6:.0f} MB')

_crew = pd.read_csv(
    os.path.join(_data_dir, 'title.crew.tsv'),
    sep='\t', na_values='\\N',
    dtype={'tconst': str, 'directors': str, 'writers': str}
)
print(f'title.crew: {len(_crew):,} rows')

_load_elapsed = time.time() - _t0_load
print(f'\nAll loaded in {_load_elapsed:.1f}s')
```

```python
# ── Cell 5: Filter to movies and merge with ratings ──
_t0 = time.time()

# Filter to movies only
_movies = _titles[_titles['titleType'] == 'movie'].copy()
print(f'Total movies: {len(_movies):,}')

# Convert startYear to numeric
_movies['startYear'] = pd.to_numeric(_movies['startYear'], errors='coerce')
_movies = _movies.dropna(subset=['startYear'])
_movies['startYear'] = _movies['startYear'].astype(int)
print(f'Movies with valid year: {len(_movies):,}')

# Merge with ratings
_movies_rated = _movies.merge(_ratings, on='tconst', how='inner')
print(f'Movies with ratings: {len(_movies_rated):,}')
print(f'Rating distribution: mean={_movies_rated["averageRating"].mean():.2f}, median={_movies_rated["averageRating"].median():.2f}')
print(f'Vote distribution: mean={_movies_rated["numVotes"].mean():,.0f}, median={_movies_rated["numVotes"].median():,.0f}')

# Focus on movies since 1950 with at least 100 votes for quality analysis
_quality_movies = _movies_rated[(_movies_rated['startYear'] >= 1950) & (_movies_rated['numVotes'] >= 100)].copy()
print(f'\nQuality movies (since 1950, 100+ votes): {len(_quality_movies):,}')
print(f'Year range: {int(_quality_movies["startYear"].min())} - {int(_quality_movies["startYear"].max())}')

_merge_elapsed = time.time() - _t0
print(f'\nFiltering & merge: {_merge_elapsed:.1f}s')
```

```python
# ── Cell 6: Genre analysis ──
_t0 = time.time()

# Parse genres (comma-separated) - avoid comprehension in loop (Issue 24)
# Split genres into rows
_genre_rows = []
for _idx in range(len(_quality_movies)):
    _row = _quality_movies.iloc[_idx]
    _g = _row['genres']
    if pd.notna(_g) and isinstance(_g, str):
        _parts = _g.split(',')
        for _p_idx in range(len(_parts)):
            _genre_rows.append({
                'tconst': _row['tconst'],
                'genre': _parts[_p_idx].strip(),
                'startYear': _row['startYear'],
                'averageRating': _row['averageRating'],
                'numVotes': _row['numVotes']
            })

_genre_df = pd.DataFrame(_genre_rows)
print(f'Genre-expanded rows: {len(_genre_df):,}')

# Count movies per genre
_genre_counts = _genre_df.groupby('genre').size().sort_values(ascending=False)
print(f'\nTop 15 genres by movie count:')
for _idx in range(min(15, len(_genre_counts))):
    _name = _genre_counts.index[_idx]
    _count = _genre_counts.iloc[_idx]
    print(f'  {_idx+1:2d}. {_name:20s} {_count:>6,} movies')

# Average rating by genre
_genre_ratings = _genre_df.groupby('genre')['averageRating'].agg(['mean', 'median', 'count'])
_genre_ratings = _genre_ratings[_genre_ratings['count'] >= 100].sort_values('mean', ascending=False)
print(f'\nHighest-rated genres (100+ movies):')
for _idx in range(min(10, len(_genre_ratings))):
    _name = _genre_ratings.index[_idx]
    _mean = _genre_ratings.iloc[_idx]['mean']
    _med = _genre_ratings.iloc[_idx]['median']
    _cnt = int(_genre_ratings.iloc[_idx]['count'])
    print(f'  {_name:20s} mean={_mean:.2f}  median={_med:.2f}  (n={_cnt:,})')

_genre_elapsed = time.time() - _t0
print(f'\nGenre analysis: {_genre_elapsed:.1f}s')
```

```python
# ── Cell 7: Genre trends over decades ──
_t0 = time.time()

# Create decade column
_genre_df['decade'] = (_genre_df['startYear'] // 10) * 10

# Count genre distribution per decade
_decade_genre = _genre_df.groupby(['decade', 'genre']).size().reset_index(name='count')
_decade_totals = _decade_genre.groupby('decade')['count'].sum().reset_index(name='total')
_decade_genre = _decade_genre.merge(_decade_totals, on='decade')
_decade_genre['pct'] = _decade_genre['count'] / _decade_genre['total'] * 100

# Track key genres across decades
_key_genres = ['Drama', 'Comedy', 'Action', 'Horror', 'Sci-Fi', 'Romance', 'Thriller', 'Documentary']
_decades = sorted(_genre_df['decade'].unique())
_decades = [_d for _d in _decades if 1950 <= _d <= 2020]

print('Genre share (%) by decade:')
print(f'{"Genre":15s}', end='')
for _idx in range(len(_decades)):
    print(f'  {_decades[_idx]}', end='')
print()
print('-' * 95)

for _g_idx in range(len(_key_genres)):
    _g = _key_genres[_g_idx]
    print(f'{_g:15s}', end='')
    for _d_idx in range(len(_decades)):
        _d = _decades[_d_idx]
        _match = _decade_genre[(_decade_genre['decade'] == _d) & (_decade_genre['genre'] == _g)]
        if len(_match) > 0:
            _pct = _match.iloc[0]['pct']
            print(f'  {_pct:5.1f}%', end='')
        else:
            print(f'     - ', end='')
    print()

# Average rating trend by decade
_decade_rating = _genre_df.groupby('decade')['averageRating'].agg(['mean', 'count'])
_decade_rating = _decade_rating[_decade_rating.index.isin(_decades)]
print(f'\nAverage rating by decade:')
for _idx in range(len(_decade_rating)):
    _d = _decade_rating.index[_idx]
    _mean = _decade_rating.iloc[_idx]['mean']
    _cnt = int(_decade_rating.iloc[_idx]['count'])
    print(f'  {_d}s: {_mean:.2f} (n={_cnt:,})')

_trend_elapsed = time.time() - _t0
print(f'\nTrend analysis: {_trend_elapsed:.1f}s')
```

```python
# ── Cell 8: Top actors and directors ──
_t0 = time.time()

# Merge principals with quality movies
_movie_principals = _principals[_principals['tconst'].isin(_quality_movies['tconst'])].copy()
print(f'Principal entries for quality movies: {len(_movie_principals):,}')

# Merge with names
_movie_principals = _movie_principals.merge(
    _names[['nconst', 'primaryName']], on='nconst', how='left'
)

# Merge with ratings
_movie_principals = _movie_principals.merge(
    _quality_movies[['tconst', 'averageRating', 'numVotes', 'startYear']],
    on='tconst', how='left'
)
print(f'Principals with names & ratings: {len(_movie_principals):,}')

# Top actors by number of quality movies
_actors = _movie_principals[_movie_principals['category'].isin(['actor', 'actress'])]
_actor_stats = _actors.groupby(['nconst', 'primaryName']).agg(
    movie_count=('tconst', 'nunique'),
    avg_rating=('averageRating', 'mean'),
    total_votes=('numVotes', 'sum'),
    career_start=('startYear', 'min'),
    career_end=('startYear', 'max')
).reset_index()
_actor_stats = _actor_stats[_actor_stats['movie_count'] >= 10].sort_values('movie_count', ascending=False)

print(f'\nTop 20 actors/actresses by movie count (10+ quality movies):')
for _idx in range(min(20, len(_actor_stats))):
    _r = _actor_stats.iloc[_idx]
    print(f'  {_idx+1:2d}. {_r["primaryName"]:30s} {int(_r["movie_count"]):3d} movies  '
          f'avg={_r["avg_rating"]:.1f}  votes={int(_r["total_votes"]):>10,}  '
          f'{int(_r["career_start"])}-{int(_r["career_end"])}')

# Top directors
_directors_principals = _movie_principals[_movie_principals['category'] == 'director']
_dir_stats = _directors_principals.groupby(['nconst', 'primaryName']).agg(
    movie_count=('tconst', 'nunique'),
    avg_rating=('averageRating', 'mean'),
    total_votes=('numVotes', 'sum')
).reset_index()
_dir_stats = _dir_stats[_dir_stats['movie_count'] >= 5].sort_values('avg_rating', ascending=False)

print(f'\nTop 20 directors by average rating (5+ quality movies):')
for _idx in range(min(20, len(_dir_stats))):
    _r = _dir_stats.iloc[_idx]
    print(f'  {_idx+1:2d}. {_r["primaryName"]:30s} {int(_r["movie_count"]):3d} movies  '
          f'avg={_r["avg_rating"]:.2f}  votes={int(_r["total_votes"]):>10,}')

_people_elapsed = time.time() - _t0
print(f'\nPeople analysis: {_people_elapsed:.1f}s')
```

```python
# ── Cell 9: Rating prediction model ──
_t0_ml = time.time()

from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import mean_absolute_error, r2_score

# Prepare features for rating prediction
# Use quality movies with genres parsed
_ml_df = _quality_movies.copy()

# Extract primary genre (first genre listed)
_ml_df['primaryGenre'] = _ml_df['genres'].apply(lambda _x: str(_x).split(',')[0] if pd.notna(_x) else 'Unknown')

# Create features
_ml_df['decade'] = (_ml_df['startYear'] // 10) * 10
_ml_df['logVotes'] = np.log1p(_ml_df['numVotes'])
_ml_df['runtimeMinutes'] = pd.to_numeric(_ml_df['runtimeMinutes'], errors='coerce')
_ml_df = _ml_df.dropna(subset=['runtimeMinutes'])
_ml_df['runtimeMinutes'] = _ml_df['runtimeMinutes'].astype(float)

# Encode categorical features
_le_genre = LabelEncoder()
_ml_df['genreCode'] = _le_genre.fit_transform(_ml_df['primaryGenre'])

_le_type = LabelEncoder()
_ml_df['isAdult'] = _ml_df['isAdult'].fillna(0).astype(int)

# Feature matrix
_feature_cols = ['startYear', 'runtimeMinutes', 'logVotes', 'genreCode', 'isAdult', 'decade']
_X = _ml_df[_feature_cols].values
_y = _ml_df['averageRating'].values

print(f'ML dataset: {len(_X):,} samples, {len(_feature_cols)} features')
print(f'Features: {_feature_cols}')
print(f'Target: averageRating (mean={_y.mean():.2f}, std={_y.std():.2f})')

_X_train, _X_test, _y_train, _y_test = train_test_split(_X, _y, test_size=0.2, random_state=42)
print(f'Train: {len(_X_train):,}, Test: {len(_X_test):,}')

# Random Forest
_rf = RandomForestRegressor(n_estimators=100, max_depth=15, random_state=42, n_jobs=-1)
_rf.fit(_X_train, _y_train)
_rf_pred = _rf.predict(_X_test)
_rf_mae = mean_absolute_error(_y_test, _rf_pred)
_rf_r2 = r2_score(_y_test, _rf_pred)
print(f'\nRandom Forest: MAE={_rf_mae:.3f}, R²={_rf_r2:.3f}')

# Gradient Boosting
_gb = GradientBoostingRegressor(n_estimators=200, max_depth=6, learning_rate=0.1, random_state=42)
_gb.fit(_X_train, _y_train)
_gb_pred = _gb.predict(_X_test)
_gb_mae = mean_absolute_error(_y_test, _gb_pred)
_gb_r2 = r2_score(_y_test, _gb_pred)
print(f'Gradient Boosting: MAE={_gb_mae:.3f}, R²={_gb_r2:.3f}')

# Feature importance
print(f'\nFeature importance (GB):')
_importances = list(zip(_feature_cols, _gb.feature_importances_))
_importances.sort(key=lambda _x: -_x[1])
for _idx in range(len(_importances)):
    _feat = _importances[_idx][0]
    _imp = _importances[_idx][1]
    print(f'  {_feat:20s} {_imp:.4f}')

_ml_elapsed = time.time() - _t0_ml
print(f'\nML pipeline: {_ml_elapsed:.1f}s')
```

```python
# ── Cell 10: Collaboration network analysis ──
_t0 = time.time()

from scipy import sparse

# Build actor co-occurrence matrix
# For each movie, all actors who appeared together form edges
_actor_movies = _actors.groupby('nconst')['tconst'].apply(list).to_dict()
_movie_actors_dict = _actors.groupby('tconst')['nconst'].apply(list).to_dict()

# Get top 1000 actors by movie count for manageable matrix
_top_actors = _actor_stats.head(1000)['nconst'].tolist()
_actor_idx = {}
for _idx in range(len(_top_actors)):
    _actor_idx[_top_actors[_idx]] = _idx

_n = len(_top_actors)
print(f'Building {_n}x{_n} co-occurrence matrix...')

# Build sparse co-occurrence matrix
_rows = []
_cols = []
_data = []

_movie_keys = list(_movie_actors_dict.keys())
for _m_idx in range(len(_movie_keys)):
    _tconst = _movie_keys[_m_idx]
    _cast = _movie_actors_dict[_tconst]
    # Filter to top actors only
    _top_cast_indices = []
    for _a_idx in range(len(_cast)):
        if _cast[_a_idx] in _actor_idx:
            _top_cast_indices.append(_actor_idx[_cast[_a_idx]])
    # Create edges between all pairs
    for _i in range(len(_top_cast_indices)):
        for _j in range(_i + 1, len(_top_cast_indices)):
            _rows.append(_top_cast_indices[_i])
            _cols.append(_top_cast_indices[_j])
            _data.append(1)
            _rows.append(_top_cast_indices[_j])
            _cols.append(_top_cast_indices[_i])
            _data.append(1)

_cooccur = sparse.csr_matrix(
    (np.array(_data), (np.array(_rows), np.array(_cols))),
    shape=(_n, _n)
)
print(f'Co-occurrence matrix: {_cooccur.shape}, {_cooccur.nnz:,} non-zero entries')

# Find most connected actors (highest degree)
_degrees = np.array(_cooccur.sum(axis=1)).flatten()
_top_connected = np.argsort(-_degrees)[:20]
print(f'\nTop 20 most connected actors:')
_top_actor_names = _top_actors
for _idx in range(20):
    _a_pos = _top_connected[_idx]
    _nconst = _top_actor_names[_a_pos]
    _name_match = _names[_names['nconst'] == _nconst]
    _aname = _name_match.iloc[0]['primaryName'] if len(_name_match) > 0 else _nconst
    _deg = int(_degrees[_a_pos])
    print(f'  {_idx+1:2d}. {_aname:30s} {_deg:5d} co-appearances')

# Find strongest collaborations (most movies together)
_cooccur_dense = _cooccur.toarray()
# Zero out diagonal
np.fill_diagonal(_cooccur_dense, 0)
# Find top pairs
_flat_indices = np.argsort(-_cooccur_dense.flatten())[:20]
_pair_rows = _flat_indices // _n
_pair_cols = _flat_indices % _n

print(f'\nTop 10 actor pairs (most movies together):')
_seen_pairs = set()
_pair_count = 0
for _idx in range(len(_pair_rows)):
    _r = _pair_rows[_idx]
    _c = _pair_cols[_idx]
    _pair_key = (min(_r, _c), max(_r, _c))
    if _pair_key in _seen_pairs:
        continue
    _seen_pairs.add(_pair_key)
    _n1 = _top_actor_names[_r]
    _n2 = _top_actor_names[_c]
    _name1 = _names[_names['nconst'] == _n1].iloc[0]['primaryName'] if len(_names[_names['nconst'] == _n1]) > 0 else _n1
    _name2 = _names[_names['nconst'] == _n2].iloc[0]['primaryName'] if len(_names[_names['nconst'] == _n2]) > 0 else _n2
    _movies_together = int(_cooccur_dense[_r, _c])
    print(f'  {_pair_count+1:2d}. {_name1} & {_name2}: {_movies_together} movies')
    _pair_count += 1
    if _pair_count >= 10:
        break

_network_elapsed = time.time() - _t0
print(f'\nNetwork analysis: {_network_elapsed:.1f}s')
```
