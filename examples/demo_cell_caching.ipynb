# Statement-Level Caching with `%cash_on`

This notebook demonstrates how `%cash_on` caches the results of each statement in every cell you run. This is useful for skipping expensive computations during interactive sessions.

```python
from cash import Cash

# Initialize Cash
cash = Cash()

# Register the magics and turn caching on for every cell below
cash.register_magic()
%cash_on
```

## Basic Caching & Output Capture

Let's define a slow computation. The first time you run this cell, it will sleep for 2 seconds. The second time, it should be instant.

Note that `print` output is also captured and replayed!

```python
import time

print("Starting slow computation...")
time.sleep(2)
if 1 == 1:
    result = 5
result += 43
print(f"Finished. Result: {result}")
```

```python
cash.explorer().show()
```

## Dependency Tracking

Cash automatically detects input variables. If an input variable changes, the cache is invalidated.

```python
multiplier = 10
```

```python
print(f"Computing with multiplier {multiplier}...")
final_value = result * multiplier
print(f"Final Value: {final_value}")
```

Try changing `multiplier` above and re-running the cell. It should re-execute.

```python
multiplier = 5
```

```python
print(f"Computing with multiplier {multiplier}...")
final_value = result * multiplier
print(f"Final Value: {final_value}")
```

## TTL (Time To Live)

You can give a statement a TTL in seconds with a `# @cash:ttl=N` comment above it (or every statement a default with `%cash_on ttl=N`).

```python
import time
# @cash:ttl=5
current_time = time.time()
print(f"Current time: {current_time}")
```
