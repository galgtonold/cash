# Statement-Level Caching with `%%cash`

This notebook demonstrates how to use the `%%cash` cell magic to cache the results of individual cells. This is useful for skipping expensive computations during interactive sessions.

```python
from cash import Cash

# Initialize Cash
cash = Cash()

# Register the magic
cash.register_magic()
```

## Basic Caching & Output Capture

Let's define a slow computation. The first time you run this cell, it will sleep for 2 seconds. The second time, it should be instant.

Note that `print` output is also captured and replayed!

```python
%%cash
import time

print("Starting slow computation...")
time.sleep(2)
if 1 == 1:
    result = 5
result += 43
print(f"Finished. Result: {result}")
```

```python
cash.explorer().show()
```

## Dependency Tracking

The magic automatically detects input variables. If an input variable changes, the cache is invalidated.

```python
multiplier = 10
```

```python
%%cash
print(f"Computing with multiplier {multiplier}...")
final_value = result * multiplier
print(f"Final Value: {final_value}")
```

Try changing `multiplier` above and re-running the cell. It should re-execute.

```python
multiplier = 5
```

```python
%%cash
print(f"Computing with multiplier {multiplier}...")
final_value = result * multiplier
print(f"Final Value: {final_value}")
```

## TTL (Time To Live)

You can specify a TTL in seconds.

```python
%%cash ttl=5
import time
current_time = time.time()
print(f"Current time: {current_time}")
```
