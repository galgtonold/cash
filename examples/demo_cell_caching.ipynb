# Statement-level caching with `%cash_on`

`%cash_on` caches each statement you run, not whole cells. Run the notebook top
to bottom once, then follow the **Try this** notes.

**Read the badge, not the clock.** cash prints a badge under every cell it runs.
Open it to see one row per statement: **CACHED** (green) means cash restored the
value instead of running the code, **EXECUTED** (ochre) means the code ran.
Printed output is no evidence either way: a restored statement replays what it
printed.

```python
import cash
%cash_on
```

## A slow statement

The first run of the next cell takes two seconds. Run it again: the badge says
CACHED for `result = slow_square(7)` and the cell returns at once. The cheap
`print` lines simply run again.

```python
import time


def slow_square(x):
    time.sleep(2)  # stands in for real work
    return x * x


print("Computing...")
result = slow_square(7)
print(f"Result: {result}")
```

## Dependencies between cells

cash records which variables each statement reads. `final_value` below reads
`result` and `multiplier`, so it is cached per value of both.

```python
multiplier = 10
```

```python
def slow_scale(x, factor):
    time.sleep(1)  # stands in for real work
    return x * factor


final_value = slow_scale(result, multiplier)
print(f"Final value: {final_value}")
```

**Try this:** change `multiplier` to `5`, run its cell, then run the cell
after it. The badge says EXECUTED, because an input changed. Now put `10` back
and run both cells again: the badge says CACHED. cash keeps one entry per set
of inputs, so going back to an earlier value costs nothing.

## Expiring a result

A `# @cash:ttl=N` comment above a statement makes its cached value expire after
N seconds. `%cash_on ttl=N` sets a default for every statement.

```python
# @cash:ttl=10
snapshot = slow_square(12)
print(snapshot)
```

**Try this:** run the cell again within ten seconds and the badge says CACHED.
Wait longer, run it again, and it says EXECUTED.
